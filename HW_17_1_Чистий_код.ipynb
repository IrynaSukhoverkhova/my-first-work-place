6.У Вас є великий DataFrame df, що представляє транзакції продажу в магазині. Цей DataFrame містить понад 1 мільйон рядків і кілька стовпців, зокрема «product_name», «product_category», «product_price» і «transaction_date».

І «product_name», і «product_category» є рядками, що представляють численні, але обмежені унікальні назви продуктів і категорії відповідно.

Напишіть код для ефективного перетворення стовпців «product_name» і «product_category» на категорійний тип для ефективного використання пам’яті.
Спробуйте вирішити це завдання, мінімізуючи використання пам'яті та зберігаючи читабельність коду. Крім того, виміряйте використання пам’яті до та після перетворення, щоб продемонструвати підвищення ефективності.

Порівняйте кількість памʼяті, яку займає наданий набір даних і новий, створений Вами. В лекції був приклад. В скільки разів менше памʼяті займає новий датафрейм?

````python
import pandas as pd
import numpy as np

np.random.seed(0)

products = ['apple', 'banana', 'cherry', 'dates', 'elderberry']
categories = ['fruit', 'fruit', 'fruit', 'fruit', 'fruit']

N_ROWS = 10**6
df = pd.DataFrame({
    'product_name': np.random.choice(products, N_ROWS),
    'product_category': np.random.choice(categories, N_ROWS),
    'product_price': np.random.uniform(1, 10, N_ROWS),
    'transaction_date': pd.date_range(start='1/1/2000', periods=N_ROWS, freq='h')
})
````

````python
mem_before = df.memory_usage(deep=True).sum()

category_cols = ["product_name", "product_category"]
df[category_cols] = df[category_cols].astype("category")

mem_after = df.memory_usage(deep=True).sum()
ratio = mem_before / mem_after

print(f"Пам'ять до перетворення: {mem_before / (1024 ** 2):.2f} МБ")
print(f"Пам'ять після перетворення: {mem_after / (1024 ** 2):.2f} МБ")
print(f"Новий датафрейм займає менше пам'яті у {ratio:.2f} раза")
````

- memory_usage(deep=True).sum() коректно враховує реальний обсяг пам'яті, який займають рядки (object) всередині DataFrame.
- перетворення виконується через df[category_cols] = df[category_cols].astype("category"), не створює додаткової копії всього датафрейму.   
- коефіцієнт mem_before / mem_after показує, у скільки разів зменшилося споживання пам'яті після перетворення рядкових стовпців з обмеженим набором унікальних значень на тип category.

````python
"""Обчислення дисперсії списку чисел."""


def calculate_variance(numbers):
    """Повертає дисперсію генеральної сукупності для списку чисел."""
    if not numbers:
        raise ValueError("Список не може бути порожнім")

    mean = sum(numbers) / len(numbers)
    squared_diffs = [(number - mean) ** 2 for number in numbers]
    return sum(squared_diffs) / len(numbers)


input_list = [75, 80, 85, 90, 95]
variance = calculate_variance(input_list)
print(f"Дисперсія: {variance}")
print(f"Стандартне відхилення: {variance ** 0.5:.2f}")
````

Що змінено щодо PEP 8:

Змінні отримали зрозумілі назви: numbers замість nl, squared_diffs замість sum_squares.
