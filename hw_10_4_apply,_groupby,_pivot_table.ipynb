0. Імпортуйте бібліотеку pandas та зчитайте дані у змінну `df` типу `pandas.DataFrame`.

```python
import pandas as pd
df = pd.read_csv('supermarket_sales.csv')
```

1. Дослідимо, який філіал супермаркету ('Branch') є найприбутковішим. Для цього знайдіть сумарний прибуток за кожним філіалом і виявіть, який філіал має найвищий.

```python
branch_profit = df.groupby('Branch')['gross income'].sum()
print("Сумарний прибуток за філіями:\n", branch_profit)
```

```python
best_branch = branch_profit.idxmax()
print(f"\nНайприбутковіший філіал: {best_branch}")
```

2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

```python
best_city = df[df['Branch'] == best_branch]['City'].iloc[0]
print(f"Філіал {best_branch} знаходиться у місті {best_city} ")
```

3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category = df.pivot_table(
    index='Branch',
    columns='Product line',
    values='Invoice ID',
    aggfunc='count'
)

print(invoices_by_category)
```

Питання 1. У який час доби (ранок, день, вечір) покупці найчастіше купують різні категорії товарів?

```python
df['Hour'] = pd.to_datetime(df['Time']).dt.hour

def time_of_day(hour):
    if 9 <= hour < 12:
        return 'Ранок (9-12)'
    elif 12 <= hour < 17:
        return 'День (12-17)'
    else:
        return 'Вечір (17+)'

df['Time_Period'] = df['Hour'].apply(time_of_day)


sales_by_time = df.pivot_table(
    index='Time_Period',
    columns='Product line',
    values='Invoice ID',
    aggfunc='count'
)

print("Розподіл покупок за часом доби:")
print(sales_by_time)
```

Для бізнесу:
9:00–12:00 - ранок є найменш прибутковим періодом для всіх категорій. Найнижчі показники зафіксовано у Health and beauty (27) та Fashion accessories (28).Відносно краще вранці продаються Home and lifestyle (39) та Electronic accessories (35).

12:00–17:00 - це найактивніший період для абсолютно всіх категорій товарів.Найвищий попит у денний час спостерігається в категоріях Sports and travel (86) та Fashion accessories (83).

Після 17:00 - ввечері активність залишається високою, особливо для Food and beverages (69), Fashion accessories (67) та Electronic accessories (66).Попит на Sports and travel суттєво падає порівняно з днем (з 86 до 49).

У ранкові години (9:00–12:00) можна зменшити кількість працівників у залі та на касах, залучаючи їх до викладки товару, інвентаризації або підготовки магазину.

Максимальну кількість касирів та консультантів варто виводити на зміну в проміжку з 12:00 до 17:00, щоб уникнути черг та втрати клієнтів.



Питання 2. В який час доби покупці ставлять найнижчі оцінки сервісу?

Питання 3. Якому способу оплати віддають перевагу постійні клієнти порівняно з звичайними покупцями і чи відрізняється їхній середній чек?

```python
payment_by_customer = df.pivot_table(
    index='Customer type',
    columns='Payment',
    values='Invoice ID',
    aggfunc='count'
)

avg_check = df.groupby('Customer type')['Total'].mean()

print("Способи оплати за типом клієнта:")
print(payment_by_customer)
print("\nСередній чек (Total):")
print(avg_check)
```

Для бізнесу: в залежності від політики компанії, можна проаналізувати, чи варто стимулювати клієнтів переходити на певний вид оплати(наприклад, робити знижки/кешбек за оплату карткою чи електрониим гаманцем або кеш), якщо це збільшує суму покупки.

Питання 4. Які категорії товарів найпопулярніші за кількістю серед чоловіків, а які - серед жінок?

```python
gender_analysis = df.pivot_table(
    index='Product line',
    columns='Gender',
    values='Invoice ID',
    aggfunc='count'
)
print(gender_analysis)
print("\n")
```

Для бізнесу: вже є певна картинка, яка дозволить ефективніше налаштувати таргетовану рекламу та планувати викладку товарів на полицях під конкретну аудиторію.

Питання 5. Товари яких категорій отримують найнижчі оцінки від чоловіків та жінок відповідно?

```python
rating_by_gender = df.pivot_table(
    index='Product line',
    columns='Gender',
    values='Rating',
    aggfunc='mean'
).round(2)

print("Середній рейтинг товарів за статтю:")
print(rating_by_gender)
```

Для бізнесу: Аудит асортименту для проблемних категорій:

Home and lifestyle (найнижчий оціночний шар у жінок): Варто переглянути якість товарів, асортимент або співвідношення ціни та якості у цій категорії. Можливо, дизайн чи практичність товарів не відповідають очікуванням жіночої аудиторії.

Health and beauty (найнижчий оціночний шар у чоловіків): Оскільки для жінок ця категорія має один із найвищих рейтингів (7.15), а для чоловіків - найнижчий (6.90), є сенс розширити саме чоловічу лінійку догляду/косметики або змінити навігацію/викладку цих товарів у магазині, щоб чоловікам було простіше обирати потрібні позиції.
