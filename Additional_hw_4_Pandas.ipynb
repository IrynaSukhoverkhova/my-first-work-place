Пропоную наступний алгоритм дій:
    1. Подивитись, які дані містяться в цій колонці. Виведіть перші і останні 5 записів. Чи однаковий формат цих даних? (спойлер: ні, бо інакше все було б просто :))
    2. Нам треба враховуючи різний формат даних перетворити рядок у колонці `Timestamp` на тип `datetime64[ns]`. Як це зробити? Для перетворення рядка на `datetime64` пропоную скористатись `datetime.datetime.strptime`.
    Для цього треба імпортувати бібліотеку `datetime`. Я пропоную написати функцію для обробки кожного окремого рядка, яка враховує довжину рядка і на основі цього застосовує різний формат даних дати.
    Як прописувати формат даних можна переглянути [тут](https://www.w3schools.com/python/python_datetime.asp).
    3. Коли функція для обробки кожного окремого рядка готова, застосуйте її для всіх даних з допомогою методу `apply`.

```python
from datetime import datetime
print("--- Перші 5 записів Timestamp ---")
print(df['Timestamp'].head())
print("\n--- Останні 5 записів Timestamp ---")
print(df['Timestamp'].tail())
```

```python
def parse_by_length(val):
    if pd.isna(val):
        return None

    val = str(val).strip()
    length = len(val)

    formats_short = ['%m/%d/%Y', '%d.%m.%Y', '%m.%d.%Y', '%Y-%m-%d']
    formats_med   = ['%m/%d/%Y %H:%M', '%d.%m.%Y %H:%M', '%m.%d.%Y %H:%M', '%Y-%m-%d %H:%M']
    formats_long  = ['%m/%d/%Y %H:%M:%S', '%d.%m.%Y %H:%M:%S', '%m.%d.%Y %H:%M:%S', '%Y-%m-%d %H:%M:%S']

    if length <= 10:
        target_formats = formats_short
    elif length <= 16:
        target_formats = formats_med
    else:
        target_formats = formats_long

    for fmt in target_formats:
        try:
            return datetime.strptime(val, fmt)
        except ValueError:
            pass

    return None
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'].apply(parse_by_length))

print("Новий тип колонки Timestamp:", df['Timestamp'].dtype)
print(df['Timestamp'].head())
```

14. Як зростає зарплата з роками? Для цього нам треба відформатувати колонку "Загальний стаж роботи за спеціальністю" аналогічно до того, як ми це робили для колонки 'Знання англійської мови' і потім вивести медіанні значення для кількості років стажу.

```python
import pandas as pd

exp_col = 'Загальний стаж роботи за спеціальністю'
salary_col = 'Зарплата у $$$ за місяць, лише ставка після сплати податків'

exp_order = [
    'Менше 3 місяців', 'Пів року', '1 рік', '1.5 року', '2 роки', '3 роки',
    '4 роки', '5 років', '6 років', '7 років', '8 років', '9 років',
    '10 років', '11 років', '12 років', '15 і більше років'
]

present_categories = [cat for cat in exp_order if cat in df[exp_col].dropna().unique()]

if not present_categories:
    present_categories = sorted(df[exp_col].dropna().unique())

df[exp_col] = pd.Categorical(df[exp_col], categories=present_categories, ordered=True)

exp_salary_median = df.groupby(exp_col, observed=False)[salary_col].median()

print("Медіанна зарплата ($) залежно від стажу:")
print(exp_salary_median)
```
