# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


```python
import scipy.stats as stats
import numpy as np

# Вхідні дані
mu_0 = 123.7  # історичне середнє
alpha_1 = 0.10  # перший рівень значущості
alpha_2 = 0.05  # другий рівень значущості
data = np.array([128, 135, 121, 142, 126, 151, 123])

# Гіпотези:
# H0: mu = 123.7 (середній час не змінився)
# H1: mu > 123.7 (середній час збільшився — односторонній (right-tailed)  t -тест)
```

```python
n = len(data)
x_bar = np.mean(data)
s = np.std(data, ddof=1)

print(f"Обсяг вибірки (n): {n}")
print(f"Вибіркове середнє (x_bar): {x_bar:.4f}")
print(f"Вибіркове стандартне відхилення (s): {s:.4f}")
```

```python
# Рахуємо t-статистику та двостороннє p-value
t_stat, p_value_two_tailed = stats.ttest_1samp(data, mu_0)

# Перетворюємо на одностороннє p-value (для альтернативи 'greater')
if x_bar > mu_0:
    p_value = p_value_two_tailed / 2
else:
    p_value = 1 - (p_value_two_tailed / 2)

df = n - 1
t_crit = stats.t.ppf(1 - alpha_1, df)

print(f"Статистика тесту (t): {t_stat:.4f}")
print(f"Критичне значення (t_crit) при alpha=0.10: {t_crit:.4f}")
print(f"p-value: {p_value:.4f}")
```

```python
if p_value < alpha_1:
    print("Рішення: Відхиляємо нульову гіпотезу (H0).")
else:
    print("Рішення: Немає підстав відхиляти нульову гіпотезу (H0).")
```

```python
print("Висновок: На рівні значущості alpha = 0.10 існують статистично значущі докази того, що після редизайну сайту середній час до покупки збільшився.")
```

```python
t_crit_05 = stats.t.ppf(1 - alpha_2, df)

print(f"Критичне значення (t_crit) при alpha=0.05: {t_crit_05:.4f}")

if p_value < alpha_2:
    print("Рішення при alpha=0.05: Відхиляємо нульову гіпотезу (H0).")
else:
    print("Рішення при alpha=0.05: НЕ відхиляємо нульову гіпотезу (H0).")

print("Чи змінилося рішення? Так, рішення змінилося: на рівні значущості 0.05 результат є статистично несуттєвим.")
```

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
import statsmodels.stats.proportion as smp
import numpy as np

# Вхідні дані згідно з умовою
nobs = 1200       # загальна кількість опитаних дорослих
count = 311       # кількість тих, хто впізнав логотип
p_null = 0.23     # гіпотетична частка до кампанії (23%)
alpha = 0.03      # рівень значущості (3%)

# Гіпотези:
# H0: p = 0.23 (частка впізнаваності не змінилася)
# H1: p > 0.23 (частка впізнаваності зросла — односторонній тест)
```

```python
sample_proportion = count / nobs

print(f"Обсяг вибірки (n): {nobs}")
print(f"Кількість успіхів (x): {count}")
print(f"Вибіркова частка (p_hat): {sample_proportion:.4f} ({sample_proportion*100:.2f}%)")
```

```python
# Виконання Z-тесту для пропорцій
z_stat, p_value = smp.proportions_ztest(
    count=count,
    nobs=nobs,
    value=p_null,
    alternative='larger'
)

print(f"Z-статистика: {z_stat:.4f}")
print(f"p-value: {p_value:.4f}")
print(f"Заданий рівень значущості (alpha): {alpha}")
```

```python
if p_value < alpha:
    print("Рішення: Відхиляємо нульову гіпотезу (H0).")
else:
    print("Рішення: Немає підстав відхиляти нульову гіпотезу (H0).")
```

**Результат:** Оскільки отримане $p$-value є меншим за $\alpha = 0.03$, ми приймаємо рішення відхилити нульову гіпотезу.

```python
print("Висновок: Рекламна кампанія була успішною — існують статистично значущі докази того, що впізнаваність логотипу компанії зросла.")
```

```python
# Довірчий інтервал для заданого рівня значущості (confidence_level = 1 - alpha = 0.97)
confint_low, confint_upp = smp.proportion_confint(
    count=count,
    nobs=nobs,
    alpha=alpha,
    method='normal'
)

confidence_level = (1 - alpha) * 100
print(f"--- Довірчий інтервал ({confidence_level:.0f}%) ---")
print(f"Нижня межа: {confint_low:.4f} ({confint_low*100:.2f}%)")
print(f"Верхня межа: {confint_upp:.4f} ({confint_upp*100:.2f}%)")

# Перевірка чи історичне значення потрапляє в інтервал
if confint_low <= p_null <= confint_upp:
    print("\nІнтерпретація: Історичне значення (23%) потрапляє в довірчий інтервал.")
else:
    print("\nІнтерпретація: Історичне значення (23%) НЕ потрапляє в довірчий інтервал, що додатково підтверджує відхилення H0 на користь зростання впізнаваності.")
```
